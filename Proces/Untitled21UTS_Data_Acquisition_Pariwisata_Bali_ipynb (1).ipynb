# Acquisition - Pengambilan Data Kunjungan Wisman ke Bali 2025

Tujuan: Mengambil 3 sumber data (Download, API, Scraping) dan menyimpannya mentah di Data/Raw/ tanpa diubah.
Email kontak: arlylelu25@gmail.com
Semua request memakai User-Agent berisi email di atas, timeout=30, dan jeda 2 detik antar-request.

```python
import requests, time, os
os.makedirs("Data/Raw", exist_ok=True)

HEADERS = {
    "User-Agent": "UTS-Data-Acquisition-Pariwisata-Bali / kontak: arlylelu25@gmail.com"
}

# Download - Data BPS (contoh link BPS)
url_download = "https://www.bps.go.id/id/statistics-table/1/kunjungan-wisman-ke-bali"
res = requests.get(url_download, headers=HEADERS, timeout=30)
time.sleep(2)

print("Download Status:", res.status_code)
# Simpan mentah
open("Data/Raw/download_mentah.html", "w", encoding="utf-8").write(res.text)
print("Tersimpan di Data/Raw/download_mentah.html")
```

## 2. Cleaning Data
Kode di bawah ini untuk membersihkan data.
Kita hapus data yang duplikat / kembar supaya hasil analisisnya tidak dobel hitung.

```python
import requests, time, json

HEADERS = {
    "User-Agent": "UTS-Data-Acquisition-Pariwisata-Bali / kontak: arlylelu25@gmail.com"
}

# API - Cuaca Bali (Open-Meteo, gratis tanpa key)
url_api = "https://api.open-meteo.com/v1/forecast?latitude=-8.4095&longitude=115.1889&current=temperature_2m,relative_humidity_2m&timezone=Asia%2FMakassar"
res = requests.get(url_api, headers=HEADERS, timeout=30)
time.sleep(2)

print("API Status:", res.status_code)
data_api = res.json()

with open("Data/Raw/api_mentah.json", "w", encoding="utf-8") as f:
    json.dump(data_api, f, indent=2, ensure_ascii=False)
print("Tersimpan di Data/Raw/api_mentah.json")
```

```python
import requests, time

HEADERS = {
    "User-Agent": "UTS-Data-Acquisition-Pariwisata-Bali / kontak: arlylelu25@gmail.com"
}

# Scraping - Daya Tarik Bali dari Wikipedia
url_scraping = "https://en.wikipedia.org/wiki/Tourism_in_Bali"
res = requests.get(url_scraping, headers=HEADERS, timeout=30)
time.sleep(2)

print("Scraping Status:", res.status_code)
open("Data/Raw/scraping_mentah.html", "w", encoding="utf-8").write(res.text)
print("Tersimpan di Data/Raw/scraping_mentah.html")
```
